```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
Vectorizer = TfidfVectorizer()
x = Vectorizer.fit_transform(df["text"])
print("original dataset shape:" , df["text"].shape)
print("numerical feature shape:" , x.shape)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression (max_iter = 1000 , random_state = 42)
model.fit(x  ,df["intent"])
```

```python
import random
```

```python
responses = (
    df.groupby("intent") ["response"]
    .apply(list)
    .to_dict()
)

print(responses["greeting"])
```

```python
def chatbot():
  print("=================")
  print("welcome to python chatbot")
  print("type 'exit' to stop the conversation")
  print("=================")

  while True:
    user_input = input("\nYou: ").strip()
    if not user_input:
      print("Bot: please type something")
      continue

    if user_input.lower() in ["exit" , "quit" , "bye" , "goodbye"]:
        print("Bot: goodbye have a nice day")
        break

    # Note: Use capital 'Vectorizer' as initialized in the previous cell
    user_features = Vectorizer.transform([user_input])
    predicted_intent = model.predict(user_features)[0]
    probabilities = model.predict_proba(user_features)[0]
    confidence = max(probabilities)

    if confidence < 0.40:
      print(
          "Bot: sorry i am not sure what you mean. "
          "please ask in a different way"
      )
      continue

    answer = random.choice(responses[predicted_intent])
    print("Bot:", answer)

chatbot()
```
